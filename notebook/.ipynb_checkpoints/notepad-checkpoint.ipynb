```python
import time
from datetime import datetime

def realtime_clock(seconds=10):
    """实时显示时间，默认显示10秒"""
    print("实时时钟 (按Ctrl+C停止)")
    print("-" * 30)
    
    try:
        for i in range(seconds):
            current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            # 使用 \r 让光标回到行首，实现原地更新
            print(f"\r当前时间: {current_time}", end="", flush=True)
            time.sleep(1)
    except KeyboardInterrupt:
        pass
    
    print("\n\n时间显示结束")

# 运行实时时钟1
realtime_clock()
```

```python
#这是一个华氏度和摄氏度之间转换的脚本
def get_valid_input():
    while True:
        try:
            user_input = input("请输入1-100的数字").strip()
            number = int(user_input)
            if not user_input:
                print("不能为空")
                continue
            if number < 1:
                print("必须大于1")
            elif number > 100:
                print("必须大于100")
            else:
                return number
        except ValueError:
            print("你就是输入错误了")
def main():
    f = get_valid_input()
    C = (f-32) * 5 / 9
    print(f"华氏度为:{f}")
    print(f"摄氏度为:{C:.2f}")


if __name__ == "__main__":
    main()


```
